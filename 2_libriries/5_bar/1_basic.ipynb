```python
import matplotlib.pyplot as plt
import numpy as np 
x=np.array([3,4,3,4,3,2])
my_label=["apple","banana","orange","Mango","Grapes","Guava"]
exp=[0.2,0.1,0.1,0.1,0.2,0.1]
plt.subplot(2,3,1)

plt.pie(x)
plt.subplot(2,3,2)

plt.title('with label')
plt.pie(x,labels=my_label)

plt.subplot(2,3,3)

plt.title('with label and set start angle ')
plt.pie(x,labels=my_label,startangle=90)

plt.subplot(2,3,4)

plt.title('with label and set start angle and the explode property ')
plt.pie(x,labels=my_label,startangle=90,explode=exp,shadow=True)


plt.legend(loc="upper right",)
plt.show()
```
