```python
import matplotlib.pyplot as plt 
import numpy as np 
x=np.array([12,34,23,44,54,23,45,21,53,15,54,32,12])
y=np.array([54,23,34,45,34,32,43,25,43,34,43,43,43])
plt.scatter(x,y,color='red')

x=np.array([12,34,43,44,52,53,35,25,73,35,44,72,72])
y=np.array([34,73,44,45,74,22,23,65,33,74,43,63,42])
plt.scatter(x,y,color='green')
plt.show()
```

we can do for different point different color 

```python
colors =np.array(["red","green","blue","yellow","pink","black","orange","purple","beige","brown","gray","cyan","magenta"])
sizes=np.array([33,23,54,53,32,24,63,43,54,23,11,12,34])
plt.scatter(x, y, c=colors,s=sizes,alpha=0.5)
plt.colorbar()

plt.show()
```
