# matplotlib is the graphical visulization library it is used to make the visual representation 

```python
import matplotlib.pyplot as plt 
import numpy as np
x=np.array([1,2,3,4,5])
y=np.array([4,5,4,3,3])
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,marker='P')
plt.show()
```

```python
plt.plot(x,y,'o-.y')
```

```python
plt.plot(x,y,'o-.y',ms=10,mfc='r',mec='b')
plt.show()
```

```python
# plt.plot(x,y,linestyle='dotted') 
plt.plot(x,y,ls=':',color='red',linewidth=3)
# this is the shorthand 
plt.show()
```
