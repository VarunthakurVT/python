we can create lables using the xlabel() and ylabel() functions

```python
import matplotlib.pyplot as plt
import numpy as np 
x=np.array([1,2,3,4,5])
y=np.array([3,4,5,2,3])
plt.plot(x,y)
plt.xlabel("distance")
plt.ylabel("time")
plt.title("distance & time ")
plt.show()
```

```python
plt.plot(x,y)
plt.grid(axis='x',color='red',linewidth='0.9',linestyle='--')
plt.show()
```
