0 D array, 1d array ,2d array ,3d array 


```python
zero=np.array(23)
one=np.array([32,34,23,32])
two=np.array([[234,43,23,43],[760,45,23,45]])
three= np.array([[[32,34,32,34],[23,43,23,54]],[[43,34,23,34],[23,34,23,45]]]) #these are the one d 2 d and 3 d array
```

```python
# to check the dimensition just do the npdim
print(zero.ndim)
print(one.ndim)
print(two.ndim)
print(three.ndim)
```

```python
# creating the 5 d array
arr5=np.array([3,4,5],ndmin=5)
print(arr5.ndim)
```

loops in array

1d

```python
for x in arr :
    print(x)
```

2d 

```python
for x in two:
    for y in x:
        print(y)
```

```python
for x in three:
    for y in x :
        for z in y:
            print(z)
```

Iterating Arrays Using nditer() this is the prebuild function 


```python
for i in np.nditer(three):
    print(i)
```
