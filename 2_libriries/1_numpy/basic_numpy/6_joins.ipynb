joining means the putting two or more then two array contents into a single array is called join <br>
In numpy is do joining by the concatenate() function  axis 0 for down and 1 for the side wase 

```python
import numpy as np 
arr1=np.array([[1,2,3],[4,5,6]])
arr2=np.array([[7,8,9],[2,3,4]])
arr=np.concat((arr1,arr2),axis=1)
arr
```

concat we have to like add manually the axis <br>
but in the Stacking is same as concatenation, the only difference is that stacking is done along a new axis.

```python
arrs=np.stack((arr1,arr2),axis=0)
arrs
```

there are another hstack and v stack h is for horizontal and v for vertical 
 d for the depth 

```python
print(np.hstack((arr1,arr2)))
print(np.vstack((arr1,arr2)))
print(np.dstack((arr1,arr2)))
```
