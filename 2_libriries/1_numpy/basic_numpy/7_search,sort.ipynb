search can be done with where() function and searchsorted()...

```python
import numpy as np 
arr=np.array([1,2,3,4,5,6,7])
s=np.where(arr==5)
s
```

```python
s=np.searchsorted(arr,5,side="right") #you can also give the multiple value
s
```

# Filtering Arrays
Getting some elements out of an existing array and creating a new array out of them is called filtering.

```python
# in that there is a concept like true of false this value add to new array this will not 
f_arr=np.array([0,1,2,3,4])
x=[True,False,False,True,False]
fn_arr=f_arr[x]
fn_arr
```

## creating the filter array

```python
filter_arr=[]
for element in fn_arr:
    if element>2:
        filter_arr.append(True)

    else:
        filter_arr.append(False)
newarr=fn_arr[filter_arr]
print(filter_arr)
newarr
```
