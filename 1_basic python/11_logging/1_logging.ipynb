Logging is recording information about what your program is doing while it runs, such as events, warnings, and errors. Python's built-in logging module does this in a more flexible way than print().

```python
import logging 
logging.debug("this is the debug message ")
logging.info("this is the info message ")
logging.warning("this is the warning message")
logging.error("this is the error message")
logging.critical("this is the critical message")
```
