```python
import json
person = {
    "name": "Aarav Sharma",
    "age": 28,
    "city": "Shimla",
    "skills": ["Python", "Data Analysis"],
    "is_employed": True
}
with open('file.json','w') as file:
    json.dump(person,file,indent=4)

```

```python
personJSON=json.dumps(person,indent=4)
```

```python
person=json.loads(personJSON)
print(person)
```
