Inheritance 


```python
class animal:
    def __init__(self,name):
        self.name=name
    def info(self):
        print(self.name)
class dog(animal):
    def sound(self):
        print(self.name,"barks")

dog1=dog("tommy")
dog1.sound()
```

super function 

```python
class animal:
    def __init__(self,name):
        self.name=name

    def info(self):
        print(self.name)
class dog(animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed=breed

dog1=dog("tommy","pitbull")
dog1.info()
```

method overriding

```python
class dept:
    def __init__(self):
        self.role="HR"
    def info(self):
        print(self.role)
class programmer(dept):
    def __init__(self):
        super().__init__()
        self.role="CFO"
p1=programmer()
p1.info()
```
