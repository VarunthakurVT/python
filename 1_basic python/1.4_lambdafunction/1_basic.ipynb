lambda function is small anonymous function written in a single line it is defined with lambda keyword<br>lambda arguments and function 

```python
add10=lambda x:x+10
print(add10(100))

mul=lambda x,y:x*y
print(mul(2,3))
```

lambda function is used when you need the simple function that only use once in your code or it is used as the argunment to the another function 
 

```python
points=[(1,2,3),(2,5,100),(3,2,3),(4,1,1)]
s_points=sorted(points,key=lambda x:x[2])
s_points
```
