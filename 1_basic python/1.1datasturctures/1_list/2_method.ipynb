### there are some method of list --<br>
| Method | Description | Example |
|---|---|---|
| `append(x)` | Adds a single item `x` to the end of the list. | `lst.append(4)` |
| `extend(iterable)` | Adds all items from an iterable (such as another list) to the end. | `lst.extend([4, 5])` |
| `insert(i, x)` | Inserts item `x` at the specified index `i`. | `lst.insert(1, 'a')` |
| `remove(x)` | Removes the first occurrence of item `x`. Raises `ValueError` if not found. | `lst.remove('a')` |
| `pop([i])` | Removes and returns the item at index `i`. Defaults to the last item. | `lst.pop(0)` |
| `clear()` | Removes all items from the list, leaving an empty list `[]`. | `lst.clear()` |
| `index(x[, start[, end]])` | Returns the index of the first occurrence of item `x`. | `lst.index(3)` |
| `count(x)` | Returns the number of times item `x` appears in the list. | `lst.count(2)` |
| `sort(*, key=None, reverse=False)` | Sorts the list in place. Can sort in ascending or descending order. | `lst.sort(reverse=True)` |
| `reverse()` | Reverses the order of elements in the list in place. | `lst.reverse()` |
| `copy()` | Returns a shallow copy of the list. | `new_lst = lst.copy()` |



```python
list=["first","second","third"] 
print("this is the initial list ",list,sep=' = ')
list.append("4th")
print("this is after i append the element in the list",list,sep=' = ')
list.extend(list)
print("this is after i use the extend",list,sep=' = ')
list.insert(4,'fifth')
print("this is after i use the insert",list,sep=' = ')
list.remove('first') #element 
print("this is after the remove this remove the first occurance ",list,sep=' = ')
list.pop(0) # this is by index it also return which element is poped 
print("after using pop ",list,sep=" = ")
list.index("first") #this return the index of the first string
list.count("4th") #this return the number of the 4th elements (it return 2)
 ## the is another method called sort it sorts the elements 
 # there is another reverse 
 
```

```python
shallow_copy=list.copy() #this one create the shallow copy 

print("this is the shallow copy ",shallow_copy)

deep_copy=list
print("this is the deep copy",deep_copy)
```
