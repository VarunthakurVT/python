Method	Syntax	What It Does
count()	tuple.count(x)	Returns how many times x appears in the tuple (0 if not found).
index()	tuple.index(x[, start[, end]])	Returns the index of the first occurrence of x (raises ValueError if not found).

```python
tuple_=("afd","varun",'a','p','p','l','e')
try:
    tuple_[0]="varun "
except TypeError:
    print("you can not change the tuple it is immutable")
if "varun" in tuple_:
    print("varun is present")
else:
    print("varun is not here ")
print(tuple_.count('p'))
print(tuple_.index('e'))
```
