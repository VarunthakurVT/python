```python
tuple_=("afd","varun",'a','p','p','l','e')
try:
    tuple_[0]="varun "
except TypeError:
    print("you can not change the tuple it is immutable")
if "varun" in tuple_:
    print("varun is present")
else:
    print("varun is not here ")
print(tuple_.count('p'))
print(tuple_.index('e'))
```
