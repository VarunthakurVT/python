# data sturcture

There is mainly four data structure
- list []
-  tuple()
- set {}
-  dictionary{ } with key and value pair



```python
l=[1,2,3,"this is list"]
type(l)
```

```python
t=(21 ,"34")
type(t)
```

```python
d={"name":"varun"}
type(d)
```

```python
s={"varun ",123}
type(s)
```

| Structure  | Brackets       | Ordered | Changeable | Duplicates          |
|------------|----------------|---------|------------|---------------------|
| List       | `[ ]`          | Yes     | Yes        | Yes                 |
| Tuple      | `( )`          | Yes     | No         | Yes                 |
| Set        | `{ }`          | No      | Yes        | No                  |
| Dictionary | `{key: value}` | Yes     | Yes        | Keys must be unique |
